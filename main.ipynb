```python
import os
import subprocess
import torch
import shutil

os.chdir("/content")
REPO_URL = "https://github.com/iam-vuvandat/ElectroMagneticPINN.git"
PROJECT_PATH = "/content/ElectroMagneticPINN/src/physics_informed_model"
MAIN_SCRIPT = "test_simulation.py"
BRANCH_NAME = "main" 

clone_target = "/content/ElectroMagneticPINN"

if torch.cuda.is_available():
    print(f"DEBUG GPU {torch.cuda.get_device_name(0)}")
else:
    print("DEBUG CPU")

if os.path.exists(clone_target):
    print(f"Đang dọn dẹp thư mục cũ: {clone_target}")
    shutil.rmtree(clone_target, ignore_errors=True)
    subprocess.run(["rm", "-rf", clone_target])

print(f"Cloning nhánh '{BRANCH_NAME}' từ {REPO_URL}")
try:
    result = subprocess.run(
        ["git", "clone", "-b", BRANCH_NAME, REPO_URL, clone_target],
        check=True,
        capture_output=True,
        text=True
    )
    print(result.stdout)
except subprocess.CalledProcessError as e:
    print(f"\n[LỖI GIT CLONE] Mã lỗi {e.returncode}:")
    print(e.stderr)
    raise e
os.chdir(PROJECT_PATH)
print(f"Working dir {os.getcwd()}")
print(f"Files {os.listdir('.')}")
get_ipython().run_line_magic("run", MAIN_SCRIPT)
```
