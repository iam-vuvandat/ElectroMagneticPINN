```python
import os
import subprocess
import torch

REPO_URL = "https://github.com/iam-vuvandat/Twin_model_pinns.git"
PROJECT_PATH = "/content/Twin_model_pinns/src/physics_informed_model"
MAIN_SCRIPT = "test_simulation.py"

clone_target = "/content/Twin_model_pinns"

if torch.cuda.is_available():
    print(f"DEBUG GPU {torch.cuda.get_device_name(0)}")
else:
    print("DEBUG CPU")

if not os.path.exists(clone_target):
    print(f"Cloning {REPO_URL}")
    subprocess.run(["git", "clone", REPO_URL, clone_target], check=True)
else:
    print(f"Pulling {clone_target}")
    subprocess.run(["git", "-C", clone_target, "pull"], check=True)

os.chdir(PROJECT_PATH)
print(f"Working dir {os.getcwd()}")
print(f"Files {os.listdir('.')}")

get_ipython().run_line_magic("run", MAIN_SCRIPT)
```
